```python
import tensorflow as tf
x =tf.Variable([[10,20,30]])
print(x.numpy())
newx = tf.concat(
    [x,[[30,40,50]]],
    axis = 0

    #axis is here for row and column
    #axis = 0 -> concatenation through row
    #axis = 1 -> concatenation through column

)
print(newx.numpy())
```
