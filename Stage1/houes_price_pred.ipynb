```python
import tensorflow as tf
from tensorflow import keras
```

```python
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense
```

```python
from tensorflow.keras.datasets import boston_housing
```

STEPS TO TRAIN THE DEEPL MODEL
# 1.Sequential
# 2.compile/preprocess
# 3.fit



```python
(x_train,y_train),(x_test, y_test) = (boston_housing.load_data())
```

```python
len(x_train[0])
# len(x_train)

#len(x_train[0]) = gives row in the database
#len(x_train)  = gives column in the databse
```

Dense are using to create a hidden layers in cnn

```python
model = Sequential()
model.add(Dense(20,activation= "relu", input_shape = (13,)))
model.add(Dense(15,activation="tanh"))
model.add(Dense(10,activation="sigmoid"))
model.add(Dense(1))
```

```python
model.compile(optimizer="Adam",loss = "mse",metrics = ["mae"])
```

```python
#after data collections
#use preprocessor data

mean = x_train.mean(axis = 0)
std_ = x_train.std(axis = 0)
```

```python
#now data is standardise

x_train = (x_train - mean)/std_
x_test = (x_test- mean)/std_
```

```python
model.fit (
    x_train,
    y_train,
    epochs=20,
    batch_size=4,
    validation_split=0.10,
    verbose=1
)
```

```python
mse_loss, mae_loss = model.evaluate(x_test,y_test)
print(mse_loss)
print(mae_loss)
```

```python
y_pred = model.predict(x_test)
```
