```python
import keras

def create_model(hp):
    model = keras.Sequential([
        # 1. Tunable Hidden Layer
        keras.layers.Dense(
            units=hp.Int("units", min_value=10, max_value=32, step=5),
            activation=hp.Choice("activation", ["relu", "tanh", "sigmoid"]),
            input_shape=(15,)
        ),
        
        # 2. Tunable Dropout
        keras.layers.Dropout(
            rate=hp.Float("dropout", min_value=0.1, max_value=0.5, step=0.15)
        ),
        
        # 3. Output Layer for Binary Classification
        keras.layers.Dense(1, activation="sigmoid")
    ])
    
    # 4. Tunable Optimizer Selection
    model.compile(
        optimizer=hp.Choice("optimizer", ["adam", "rmsprop", "sgd"]),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )
    
    return model
```

```python
import keras_tuner as kt
import shutil
import os

# 1. Automatically wipe out the old corrupted trial cache folder
if os.path.exists("my_tuner_workspace"):
    shutil.rmtree("my_tuner_workspace")

# 2. Re-instantiate the tuner cleanly (Fix variable naming vs directory strings)
tuner = kt.RandomSearch(
    create_model, 
    objective="val_accuracy", 
    max_trials=5, 
    directory="my_tuner_workspace",  # Brand new clean directory path
    overwrite=True                   # Forces a fresh search workspace
)

# 3. Restart the tuning search
tuner.search(X_train, Y_train, epochs=10, validation_split=0.10)
```

```python
# Evaluate model performance on unseen test data
test_loss, test_accuracy = best_model.evaluate(X_test, Y_test)

print("\n--- FINAL TEST SET RESULTS ---")
print(f"Test Loss:     {test_loss:.4f}")
print(f"Test Accuracy: {test_accuracy * 100:.2f}%")
```
