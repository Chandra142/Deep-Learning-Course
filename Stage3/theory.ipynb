## Image Segmentation Theory

These notes summarise the task of assigning a class or object identity to image regions, along with widely used segmentation architectures.

## Segmentation types

- **Semantic segmentation** assigns a class label to every pixel. All objects from the same class share one label.
- **Instance segmentation** identifies each object separately, even when multiple objects belong to the same class.
- **Panoptic segmentation** combines semantic and instance segmentation: every pixel receives a class label and countable objects receive an instance ID.

## Edge detection

Edges are locations with a sharp change in image intensity or colour. They help define object boundaries.

- **Continuity:** connected edge pixels can describe a boundary.
- **Discontinuity:** abrupt changes in intensity, colour, or texture often indicate an edge.

Common operators include Sobel, Prewitt, and Canny.

## Clustering-based segmentation

Clustering groups pixels with similar features, such as colour, intensity, texture, or position. K-means is a simple example. It can provide a quick baseline, but it does not naturally capture object shape or high-level context.

## Encoder-decoder networks

Many segmentation models use two main parts:

1. **Encoder:** progressively downsamples the image to learn compact, high-level features.
2. **Decoder:** upsamples those features to recover a dense, pixel-level prediction.

Downsampling expands the receptive field and reduces computation; upsampling restores spatial resolution.

## U-Net

U-Net is an encoder-decoder architecture with skip connections between matching encoder and decoder levels. The skip connections preserve fine spatial detail that could otherwise be lost during downsampling. U-Net is especially common in medical-image segmentation.

## Fully Convolutional Networks (FCNs)

An FCN replaces fully connected classification layers with convolutional layers, allowing the network to produce a spatial output map rather than one label for the entire image. Transposed convolutions or interpolation are then used to upsample predictions to the input resolution.

## SegNet

SegNet is an encoder-decoder segmentation architecture. Its decoder reuses the pooling indices recorded by the encoder to guide upsampling. This helps preserve boundary information while keeping the decoder relatively memory efficient.

## Mask R-CNN

Mask R-CNN extends Faster R-CNN for instance segmentation. For every detected object, it predicts a class, a bounding box, and a pixel mask. It is useful when individual objects must be detected and segmented separately.

## Vision Transformers (ViT)

Vision Transformers divide an image into patches, convert them to embeddings, and use self-attention to model relationships between patches. For segmentation, transformer-based encoders are usually paired with a decoder that produces dense pixel predictions. Self-attention can capture long-range context that is difficult for local convolutions alone.

## Key takeaways

- Choose semantic segmentation when only a class map is needed.
- Choose instance segmentation when separate object masks are required.
- Encoder-decoder designs balance semantic understanding with spatial precision.
- Architecture choice depends on accuracy needs, available compute, memory budget, and the type of annotation available.
